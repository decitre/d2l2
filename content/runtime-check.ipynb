Reports the kernel and checks that NumPy runs. Adapted D2L chapters will sit next to this notebook.

```python
import platform, sys
from importlib.util import find_spec

import numpy as np

kind = "xeus-python" if find_spec("pyjs") else "pyodide"  # xeus-python also has a pyodide module
x = np.arange(12).reshape(3, 4)
assert x.sum() == 66
print(f"{kind} on {sys.platform}, Python {platform.python_version()}, NumPy {np.__version__}")
```
